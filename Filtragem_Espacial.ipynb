```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.util import random_noise
```

```python
img = cv2.imread('data/Fig0333(a)(test_pattern_blurring_orig).tif', 1)

if img is None:
    print("Erro ao carregar a imagem")
else:
    print("Imagem carregada com sucesso:", img.shape)

cv2.imshow('in',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Conversão básica

B, G, R = cv2.split(img)
img_grayscale_basic = (B + G + R)/3

img_grayscale_basic = np.array(img_grayscale_basic, dtype = np.uint8)

cv2.imshow('img_grayscale_basic',img_grayscale_basic)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Conversão ponderada

B, G, R = cv2.split(img)
img_grayscale_pondered = 0.299*R + 0.587*G + 0.114*B

img_grayscale_pondered = np.array(img_grayscale_pondered, dtype=np.uint8)

cv2.imshow('img_grayscale_basic', img_grayscale_pondered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Função Gauss Kernel
def gauss_create(sigma=1, size_x=3, size_y=3):
    '''
    Create normal (gaussian) distribuiton
    '''
    x, y = np.meshgrid(np.linspace(-1,1,size_x), np.linspace(-1,1,size_y))
    calc = 1/((2*np.pi*(sigma**2)))
    exp = np.exp(-(((x**2) + (y**2))/(2*(sigma**2))))

    return exp*calc
```

```python
def mean_kernel(size):
    kernel = np.ones((size, size), dtype=float)
    kernel = kernel / np.sum(kernel)
    return kernel
```

```python
def add_padding(img, padding_height, padding_width):
    n, m = img.shape

    padded_img = np.zeros((n + padding_height * 2, m + padding_width * 2))
    padded_img[padding_height : n + padding_height, padding_width : m + padding_width] = img

    return padded_img
```

```python
def conv2d(img, kernel, padding=True):
    # Get dimensions of the kernel
    k_height, k_width = kernel.shape  # Atribui valor à variável k_height, k_width

    # Get dimensions of the image
    img_height, img_width = img.shape  # Atribui valor à variável img_height, img_width

    # Calculate padding required
    pad_height = k_height // 2  # Atribui valor à variável pad_height
    pad_width = k_width // 2  # Atribui valor à variável pad_width

    # Create a padded version of the image to handle edges
    if padding == True:
        padded_img = add_padding(img, pad_height, pad_width)  # Atribui valor à variável padded_img

    # print(padded_img)

    # Initialize an output image with zeros
    output = np.zeros((img_height, img_width), dtype=float)  # Atribui valor à variável output


    # Perform convolution (Implementação)
    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + k_height, j:j + k_width]
            value = np.sum(region * kernel)
            output[i, j] = value

    output = np.clip(output, 0, 255)

    return np.array(output, dtype=np.uint8)
```

```python
def median_filter(img, size, padding=True):
    img_height, img_width = img.shape
    pad = size // 2

    if padding == True:
        padded_img = add_padding(img, pad, pad)
    else:
        padded_img = img

    output = np.zeros((img_height, img_width), dtype=float)

    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + size, j:j + size]
            values = region.flatten()
            values.sort()
            median_value = values[len(values) // 2]
            output[i, j] = median_value

    return np.array(output, dtype=np.uint8)
```

## Máscaras

### Gaussiano
Suaviza a imagem dando mais peso aos pixels próximos.

- 3x3
- 5x5
- 7x7

### Média
Calcula a média dos pixels vizinhos.

- 3x3
- 5x5
- 7x7

### Mediana
Usa a mediana dos pixels vizinhos, sendo bom para remover ruídos.

- 3x3
- 5x5
- 7x7

```python
img = img_grayscale_pondered
mean_3 = conv2d(img, mean_kernel(3))
mean_5 = conv2d(img, mean_kernel(5))
mean_7 = conv2d(img, mean_kernel(7))

gauss_3 = conv2d(img, gauss_create(sigma=1.0, size_x=3, size_y=3))
gauss_5 = conv2d(img, gauss_create(sigma=1.2, size_x=5, size_y=5))
gauss_7 = conv2d(img, gauss_create(sigma=1.5, size_x=7, size_y=7))

median_3 = median_filter(img, 3)
median_5 = median_filter(img, 5)
median_7 = median_filter(img, 7)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(mean_3, cmap='gray')
plt.title('Média 3x3')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gauss_3, cmap='gray')
plt.title('Gaussiano 3x3')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(median_3, cmap='gray')
plt.title('Mediana 3x3')
plt.axis('off')

plt.show()
```

```python
img_sp = random_noise(img, mode='s&p', amount=0.05)
img_sp = (img_sp * 255).astype(np.uint8)

img_gn = random_noise(img, mode='gaussian', var=0.01)
img_gn = (img_gn * 255).astype(np.uint8)

mean_sp_3 = conv2d(img_sp, mean_kernel(3))
gauss_sp_3 = conv2d(img_sp, gauss_create(1.0, 3, 3))
median_sp_3 = median_filter(img_sp, 3)

mean_gn_3 = conv2d(img_gn, mean_kernel(3))
gauss_gn_3 = conv2d(img_gn, gauss_create(1.0, 3, 3))
median_gn_3 = median_filter(img_gn, 3)
```
