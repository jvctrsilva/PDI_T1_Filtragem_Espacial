```python
#Conversão básica

def grayscale_basic(img):
    B, G, R = cv2.split(img)
    img_grayscale_basic = (B + G + R)/3

    img_grayscale_basic = np.array(img_grayscale_basic, dtype = np.uint8)

    cv2.imshow('img_grayscale_basic',img_grayscale_basic)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#Conversão ponderada

def grayscale_pondered(img):
    B, G, R = cv2.split(img)
    img_grayscale_pondered = 0.299*R + 0.587*G + 0.114*B

    img_grayscale_pondered = np.array(img_grayscale_pondered, dtype=np.uint8)

    return img_grayscale_pondered
```

```python
#Função Gauss Kernel

def gauss_create(sigma=1, size_x=3, size_y=3):
    x, y = np.meshgrid(np.linspace(-1,1,size_x), np.linspace(-1,1,size_y))
    calc = 1/(2*np.pi*(sigma**2))
    exp = np.exp(-(((x**2) + (y**2))/(2*(sigma**2))))

    return exp*calc
```

```python
#Função kernel média

def mean_kernel(size):
    kernel = np.ones((size, size), dtype=float)
    kernel = kernel / np.sum(kernel)
    return kernel
```

```python
# Função adiciona borda

def add_padding(img, padding_height, padding_width):
    n, m = img.shape

    padded_img = np.zeros((n + padding_height * 2, m + padding_width * 2))
    padded_img[padding_height : n + padding_height, padding_width : m + padding_width] = img

    return padded_img
```

```python
#Função Kernel Mediana

def median_filter(img, size, padding=True):
    img_height, img_width = img.shape
    pad = size // 2

    if padding == True:
        padded_img = add_padding(img, pad, pad)
    else:
        padded_img = img

    output = np.zeros((img_height, img_width), dtype=float)

    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + size, j:j + size]
            values = region.flatten()
            values.sort()
            median_value = values[len(values) // 2]
            output[i, j] = median_value

    return np.array(output, dtype=np.uint8)
```

```python
#Função de convolução

def conv2d(img, kernel, padding=True):
    # Get dimensions of the kernel
    k_height, k_width = kernel.shape  # Atribui valor à variável k_height, k_width

    # Get dimensions of the image
    img_height, img_width = img.shape  # Atribui valor à variável img_height, img_width

    # Calculate padding required
    pad_height = k_height // 2  # Atribui valor à variável pad_height
    pad_width = k_width // 2  # Atribui valor à variável pad_width

    # Create a padded version of the image to handle edges
    if padding == True:
        padded_img = add_padding(img, pad_height, pad_width)  # Atribui valor à variável padded_img

    # print(padded_img)

    # Initialize an output image with zeros
    output = np.zeros((img_height, img_width), dtype=float)  # Atribui valor à variável output


    # Perform convolution (Implementação)
    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + k_height, j:j + k_width]
            value = np.sum(region * kernel)
            output[i, j] = value

    output = np.clip(output, 0, 255)

    return np.array(output, dtype=np.uint8)
```

## Observações sobre os filtros de suavização

De acordo com os dois testes realizados (sem adição de ruído e com adição), podemos observar a seguinte situação:

### Filtro da Média
- **Comportamento visual:** Suaviza a imagem uniformemente, mas tende a **borrar bordas e detalhes finos**, especialmente com kernels maiores (5x5 e 7x7).
- **Ruído:** Remove parcialmente ruídos leves, mas não é eficaz contra ruídos impulsivos ("sal e pimenta").
- **Esmaecimento:** Quanto maior o kernel, mais a imagem parece "esmaecida" ou desfocada, pois cada pixel é substituído pela média dos vizinhos.

### Filtro Gaussiano
- **Comportamento visual:** Produz uma suavização mais natural que a média, **preservando melhor as bordas**, pois dá mais peso aos pixels próximos do centro do kernel.
- **Ruído:** Reduz ruídos leves de forma eficaz, mas ainda pode suavizar detalhes pequenos.
- **Esmaecimento:** O aumento do sigma ou do tamanho do kernel aumenta o efeito de desfoque, mas de maneira mais gradual e perceptualmente agradável que a média.

### Filtro da Mediana
- **Comportamento visual:** Mantém bordas e estruturas mais nítidas mesmo com kernels maiores.
- **Ruído:** Excelente para remover **ruídos impulsivos**, sem borrar demais a imagem.
- **Esmaecimento:** Comparado à média e ao gaussiano, a mediana causa menos esmaecimento; a imagem permanece visualmente mais definida, mesmo para kernels 5x5 e 7x7.

### Resumo do comportamento
- **Média:** suavização uniforme, borramento significativo das bordas, boa para ruído leve.
- **Gaussiano:** suavização ponderada, preserva bordas melhor que a média, reduz ruído leve com aparência natural.
- **Mediana:** preserva bordas, excelente contra ruído impulsivo, mantém a definição da imagem.

Em geral, o tamanho do kernel amplifica os efeitos descritos: **kernels maiores aumentam a suavização e o esmaecimento**, enquanto kernels menores mantém mais detalhes.

```python
#Função para normalizar mínimo e máximo
def normalize_minmax(img):
    img = img.astype(np.float32)
    min_val = img.min()
    max_val = img.max()

    if max_val == min_val:
        return np.zeros_like(img, dtype=np.uint8)

    norm = (img - min_val) * 255.0 / (max_val - min_val)
    return norm.astype(np.uint8)
```

```python
#Kernel
laplacian_1 = np.array([
    [0, 1, 0],
    [1, -4, 1],
    [0, 1, 0]
], dtype=float)

laplacian_2 = np.array([
    [1, 1, 1],
    [1, -8, 1],
    [1, 1, 1]
], dtype=float)

laplacian_3 = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
], dtype=float)

laplacian_4 = np.array([
    [-1, -1, -1],
    [-1, 8, -1],
    [-1, -1, -1]
], dtype=float)
```

```python
#Função Laplacian
def laplacian_sharpen(img, kernel):
    laplacian = conv2d(img, kernel)

    if kernel[kernel.shape[0]//2, kernel.shape[1]//2] < 0:
        sharpen = img.astype(np.float32) - laplacian
    else:
        sharpen = img.astype(np.float32) + laplacian

    sharpen = np.clip(sharpen, 0, 255)
    return laplacian, sharpen.astype(np.uint8)
```

```python
#Função Unsharp Masking
def unsharp_masking(img, blur_kernel):
    img = img.astype(np.float32)
    blurred = conv2d(img, blur_kernel).astype(np.float32)

    mask = img - blurred
    sharpened = img + mask

    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    return blurred, mask, sharpened
```

Highboost Filtering

```python
#Função highboost filtering

def highboost_filter(img, blur_kernel, k=1.5):
    img = img.astype(np.float32)
    blurred = conv2d(img, blur_kernel).astype(np.float32)

    mask = img - blurred
    sharpened = img + k * mask

    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    return blurred, mask, sharpened
```

```python
#Função borda roberts

def roberts_edge(img):
    gx_kernel = np.array([
        [-1, 0],
        [0, 1]
    ], dtype=np.float32)

    gy_kernel = np.array([
        [0, -1],
        [1, 0]
    ], dtype=np.float32)

    gx = conv2d(img, gx_kernel, padding=True)
    gy = conv2d(img, gy_kernel, padding=True)

    magnitude = np.sqrt(gx**2 + gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)

    return gx, gy, magnitude
```

```python
# Função borda sobel

def sobel_edge(img):
    gx_kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)

    gy_kernel = np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ], dtype=np.float32)

    gx = conv2d(img, gx_kernel, padding=True)
    gy = conv2d(img, gy_kernel, padding=True)

    magnitude = np.sqrt(gx**2 + gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)

    return gx, gy, magnitude
```

```python
#Função borda laplacian
def laplacian_edge(img, kernel):
    lap = conv2d(img, kernel, padding=True)
    return lap, normalize_minmax(lap)
```

## Observações sobre os filtros de sharpening e detectores de borda

Com base nos testes realizados com os filtros de realce (sharpening) e detectores de borda, é possível observar diferenças claras no comportamento visual das imagens:

### Filtro Laplaciano (Sharpening e detecção)
- *Comportamento visual:* Destaca bem as bordas e as mudanças mais bruscas da imagem, deixando ela com mais detalhes, mas com um aspecto mais “duro” e menos natural.
- *Ruído:* Amplifica significativamente o ruído, já que também realça variações pequenas de intensidade.
- *Esmaecimento:* Não suaviza a imagem; pelo contrário, pode gerar aparência mais granulada.

### Sharpening (Laplaciano aplicado na imagem original)
- *Comportamento visual*: Melhora a nitidez geral da imagem, tornando contornos mais definidos sem perder completamente a informação original.
- *Ruído:* Ainda amplifica ruídos, mas de forma menos agressiva que o Laplaciano puro.
- *Esmaecimento:* Reduz o aspecto borrado da imagem, recuperando detalhes.

### Unsharp Masking e Highboost Filtering
- *Comportamento visual:* Realçam detalhes finos e bordas de forma mais controlada e natural.
- *Ruído:* Amplificam ruído, porém de maneira proporcional ao ganho
- *Esmaecimento:* Reduzem significativamente o desfoque, mantendo aparência mais suave que o Laplaciano.

## Detectores de borda

### Roberts
- *Comportamento visual:* Detecta bordas rapidamente, porém com baixa precisão e maior sensibilidade a ruído.
- *Ruído:* Muito sensível, gerando bordas falsas.
- *Esmaecimento:* Não trata suavização, resultando em detecção instável.
### Sobel
- *Comportamento visual:* Detecta bordas com mais suavidade e continuidade que Roberts.
- *Ruído:* Menos sensível ao ruído devido ao efeito de suavização embutido.
- *Esmaecimento:* Mantém bom equilíbrio entre detecção e suavidade.

### Laplaciano (detector)
- *Comportamento visual:* Detecta bordas em todas as direções, mas pode gerar bordas duplas.
- *Ruído:* Muito sensível ao ruído.
- *Esmaecimento:* Não realiza suavização, o que prejudica o resultado.
### Canny
- *Comportamento visual:* Produz bordas finas, contínuas e bem definidas.
- *Ruído:* Muito resistente ao ruído devido ao uso de suavização (Gaussian blur).
- *Esmaecimento:* Controlado internamente, preservando detalhes importantes.

## Observações técnicas – Sobel pós-suavização vs Canny
Após a aplicação de um filtro de suavização antes do operador de Sobel, observa-se uma redução significativa dos ruídos de alta frequência, o que diminui a ocorrência de bordas falsas e torna a detecção mais estável em comparação ao Sobel aplicado diretamente. Esse pré-processamento melhora a qualidade geral do gradiente calculado.
Mesmo assim, quando comparado ao Canny:
- As bordas obtidas com Sobel ainda são mais espessas e menos bem definidas, devido à ausência de etapas de refinamento
- Pode haver descontinuidades, principalmente em regiões de baixo contraste ou transições suaves
- Ainda existe certa sensibilidade a ruídos residuais, mesmo após a suavização
- Apresenta menor precisão na localização das bordas, com respostas mais difusas

Por outro lado, o Canny produz bordas mais finas, contínuas e bem localizadas, pois utiliza etapas adicionais como supressão de não-máximos e histerese. Isso resulta em uma detecção mais robusta, precisa e consistente, especialmente em imagens com variações de intensidade e presença de ruído.
