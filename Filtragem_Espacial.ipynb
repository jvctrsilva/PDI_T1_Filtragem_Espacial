```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.util import random_noise
```

```python
img = cv2.imread('data/Fig0338(a)(blurry_moon).tif', 1)

if img is None:
    print("Erro ao carregar a imagem")
else:
    print("Imagem carregada com sucesso:", img.shape)

cv2.imshow('in',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Conversão básica

B, G, R = cv2.split(img)
img_grayscale_basic = (B + G + R)/3

img_grayscale_basic = np.array(img_grayscale_basic, dtype = np.uint8)

cv2.imshow('img_grayscale_basic',img_grayscale_basic)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Conversão ponderada

B, G, R = cv2.split(img)
img_grayscale_pondered = 0.299*R + 0.587*G + 0.114*B

img_grayscale_pondered = np.array(img_grayscale_pondered, dtype=np.uint8)

cv2.imshow('img_grayscale_basic', img_grayscale_pondered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Função Gauss Kernel
def gauss_create(sigma=1, size_x=3, size_y=3):
    '''
    Create normal (gaussian) distribuiton
    '''
    x, y = np.meshgrid(np.linspace(-1,1,size_x), np.linspace(-1,1,size_y))
    calc = 1/((2*np.pi*(sigma**2)))
    exp = np.exp(-(((x**2) + (y**2))/(2*(sigma**2))))

    return exp*calc
```

```python
def mean_kernel(size):
    kernel = np.ones((size, size), dtype=float)
    kernel = kernel / np.sum(kernel)
    return kernel
```

```python
def add_padding(img, padding_height, padding_width):
    n, m = img.shape

    padded_img = np.zeros((n + padding_height * 2, m + padding_width * 2))
    padded_img[padding_height : n + padding_height, padding_width : m + padding_width] = img

    return padded_img
```

```python
def median_filter(img, size, padding=True):
    img_height, img_width = img.shape
    pad = size // 2

    if padding == True:
        padded_img = add_padding(img, pad, pad)
    else:
        padded_img = img

    output = np.zeros((img_height, img_width), dtype=float)

    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + size, j:j + size]
            values = region.flatten()
            values.sort()
            median_value = values[len(values) // 2]
            output[i, j] = median_value

    return np.array(output, dtype=np.uint8)
```

```python
def conv2d(img, kernel, padding=True):
    # Get dimensions of the kernel
    k_height, k_width = kernel.shape  # Atribui valor à variável k_height, k_width

    # Get dimensions of the image
    img_height, img_width = img.shape  # Atribui valor à variável img_height, img_width

    # Calculate padding required
    pad_height = k_height // 2  # Atribui valor à variável pad_height
    pad_width = k_width // 2  # Atribui valor à variável pad_width

    # Create a padded version of the image to handle edges
    if padding == True:
        padded_img = add_padding(img, pad_height, pad_width)  # Atribui valor à variável padded_img

    # print(padded_img)

    # Initialize an output image with zeros
    output = np.zeros((img_height, img_width), dtype=float)  # Atribui valor à variável output


    # Perform convolution (Implementação)
    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + k_height, j:j + k_width]
            value = np.sum(region * kernel)
            output[i, j] = value

    output = np.clip(output, 0, 255)

    return np.array(output, dtype=np.uint8)
```

```python
img = img_grayscale_pondered
mean_3 = conv2d(img, mean_kernel(3))
mean_5 = conv2d(img, mean_kernel(5))
mean_7 = conv2d(img, mean_kernel(7))

gauss_3 = conv2d(img, gauss_create(sigma=1.0, size_x=3, size_y=3))
gauss_5 = conv2d(img, gauss_create(sigma=1.2, size_x=5, size_y=5))
gauss_7 = conv2d(img, gauss_create(sigma=1.5, size_x=7, size_y=7))

median_3 = median_filter(img, 3)
median_5 = median_filter(img, 5)
median_7 = median_filter(img, 7)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(mean_3, cmap='gray')
plt.title('Média 3x3')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gauss_3, cmap='gray')
plt.title('Gaussiano 3x3')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(median_3, cmap='gray')
plt.title('Mediana 3x3')
plt.axis('off')

plt.show()
```

```python
# Imagem com ruído
img_sp = random_noise(img, mode='s&p', amount=0.05)
img_sp = (img_sp * 255).astype(np.uint8)

img_gn = random_noise(img, mode='gaussian', var=0.01)
img_gn = (img_gn * 255).astype(np.uint8)

mean_sp_3 = conv2d(img_sp, mean_kernel(7))
gauss_sp_3 = conv2d(img_sp, gauss_create(1.0, 7, 7))
median_sp_3 = median_filter(img_sp, 7)

mean_gn_3 = conv2d(img_gn, mean_kernel(7))
gauss_gn_3 = conv2d(img_gn, gauss_create(1.0, 7, 7))
median_gn_3 = median_filter(img_gn, 7)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img_sp, cmap='gray')
plt.title('Ruído Sal e Pimenta')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(mean_sp_3, cmap='gray')
plt.title('Média 3x3')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gauss_sp_3, cmap='gray')
plt.title('Gaussiano 3x3')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(median_sp_3, cmap='gray')
plt.title('Mediana 3x3')
plt.axis('off')

plt.show()
```

```python
def conv2d_float(img, kernel, padding=True):
    k_height, k_width = kernel.shape
    img_height, img_width = img.shape

    pad_height = k_height // 2
    pad_width = k_width // 2

    if padding:
        padded_img = add_padding(img, pad_height, pad_width)
    else:
        padded_img = img

    output = np.zeros((img_height, img_width), dtype=float)

    for i in range(img_height):
        for j in range(img_width):
            region = padded_img[i:i + k_height, j:j + k_width]
            output[i, j] = np.sum(region * kernel)

    return output

def normalize_image(img):
    img = np.abs(img)
    img = np.clip(img, 0, 255)
    return img.astype(np.uint8)
```

```python
laplacian_1 = np.array([
    [0, 1, 0],
    [1, 4, 1],
    [0, 1, 0]
], dtype=float)

laplacian_2 = np.array([
    [1, 1, 1],
    [1, -8, 1],
    [1, 1, 1]
], dtype=float)

laplacian_3 = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
], dtype=float)

laplacian_4 = np.array([
    [-1, -1, -1],
    [-1, 8, -1],
    [-1, -1, -1]
], dtype=float)
```

```python
# Laplacian

kernels = {
    "Laplacian 4-neighbors": laplacian_1,
    "Laplacian 8-neighbors (invertido)": laplacian_2,
    "Laplacian 4-neighbors (invertido)": laplacian_3,
    "Laplacian 8-neighbors": laplacian_4
}

img = img_grayscale_pondered

plt.figure(figsize=(18, 8))

for i, (name, kernel) in enumerate(kernels.items()):

    lap_img = conv2d_float(img, kernel)
    lap_img_show = normalize_image(lap_img)

    sharp_lap = img.astype(float) - lap_img
    sharp_lap = np.clip(sharp_lap, 0, 255).astype(np.uint8)

    plt.subplot(2, 4, i + 1)
    plt.imshow(lap_img_show, cmap='gray')
    plt.title(f"{name} - Laplacian")
    plt.axis('off')

    plt.subplot(2, 4, i + 5)
    plt.imshow(sharp_lap, cmap='gray')
    plt.title(f"{name} - Sharpen")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
roberts_x = np.array([
    [1, 0],
    [0, -1]
], dtype=float)

roberts_y = np.array([
    [0, 1],
    [-1, 0]
], dtype=float)

sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=float)

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=float)
```

```python
lap_edges = conv2d_float(img, laplacian_2)
lap_edges_show = normalize_image(lap_edges)

sharp_lap = img.astype(float) - lap_edges
sharp_lap = np.clip(sharp_lap, 0, 255).astype(np.uint8)

sharp_direct = conv2d(img, laplacian_2)
```

```python
blurred = conv2d(img, gauss_create(sigma=1.0, size_x=5, size_y=5))

mask = img.astype(float) - blurred.astype(float)

unsharp = img.astype(float) + mask
unsharp = np.clip(unsharp, 0, 255).astype(np.uint8)

k = 1.8
highboost = img.astype(float) + k * mask
highboost = np.clip(highboost, 0, 255).astype(np.uint8)
```

```python
rx = conv2d_float(img, roberts_x)
ry = conv2d_float(img, roberts_y)

roberts = np.sqrt(rx**2 + ry**2)
roberts = normalize_image(roberts)
```

```python
sx = conv2d_float(img, sobel_x)
sy = conv2d_float(img, sobel_y)

sobel = np.sqrt(sx**2 + sy**2)
sobel = normalize_image(sobel)
```

```python
canny = cv2.Canny(img, 100, 200)
```

```python
img_smooth = conv2d(img, gauss_create(sigma=1.0, size_x=5, size_y=5))

sx_s = conv2d_float(img_smooth, sobel_x)
sy_s = conv2d_float(img_smooth, sobel_y)

sobel_smooth = np.sqrt(sx_s**2 + sy_s**2)
sobel_smooth = normalize_image(sobel_smooth)
```

```python
plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(lap_edges_show, cmap='gray')
plt.title('Laplaciano')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(sharp_direct, cmap='gray')
plt.title('Sharpening')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(unsharp, cmap='gray')
plt.title('Unsharp')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(highboost, cmap='gray')
plt.title('Highboost')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(sobel, cmap='gray')
plt.title('Sobel')
plt.axis('off')

plt.show()
```
